# Figures S1-S5 of the Supporting Information

The tables of the Supporting Information are written by `scripts/03_evaluate.py` to `results/tables/` and, as
text, to `results/summary.txt`.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
%matplotlib inline
from swot2argo import data, models, evaluate, gradients, plots
pd.set_option("display.precision", 3); pd.set_option("display.width", 160)
T = data.tests()
```

```python
P, B, D = data.predictions(), data.benchmarks(), data.gradients()
```

**Figure S1.** (a-d) The observed density anomaly at mooring S1 on four days and its reconstruction from 2 and 4
modes; (e-g) the fraction of the variance of the observed horizontal gradient between mooring pairs explained by the
observed profiles truncated to 1-4 modes, by separation, in the mixed-layer interior, at its base and at 75 m.

```python
fig = plots.pca_gradient_modes(T, data.history().anom); plt.show()
```

**Figure S2.** Every S2 profile of the science-orbit year: observed, the network with each input, the
climatology, GLORYS12, ARMOR3D and HYCOM.

```python
fig = plots.s2_daily(P, T, B); plt.show()
```

**Figure S3.** The horizontal density gradient between mooring pairs, by separation: the mean magnitude
against depth for each estimate, and the distribution of the signed gradient in the mixed-layer interior and at its
base.

```python
fig = plots.gradients_by_scale(P, T, B); plt.show()
```

**Figure S4.** The gliders: density anomaly along the two longest transects of ru32, observed and from each
estimate, and the correlation of the signed mixed-layer gradient by separation.

```python
fig = plots.gliders(P, T, D, B); plt.show()
```

**Figure S5.** SWOT's profile gain against the time between the observed profile and the pass: (a) the mooring
array and S2, against the hourly record before and after the pass; (b) the gliders and the Argo pairs (out-of-fold
predictions of the selection, `results/cv_predictions.npz`).

```python
cv = data.cv_predictions()
fig = plots.tide_lag(P, T, data.pairs(), (cv["pretrained|ssh,sst,mdt|4||swot"], cv["pretrained|ssh,sst,mdt|4||duacs"])); plt.show()
```
